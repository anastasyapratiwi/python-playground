# Step 1 
Declare variables that store information about the user and the movie show.

```python
base_price = 15
age = 21
```

# Step 2
Store some string values about the movie ticket:
1. The type of seat the user has selected
2. The show time of the movie

```python
seat_type = "Gold"
show_time = "Evening"
```

# Step 3
Check if the user is eligible to book a movie ticket based on their age.

```python
if age > 17:
    print("User is eligible to book a ticket")
```

# Step 4
Check whether the user is allowed to book an evening show based on their age.

```python
if age >= 21:
    print("User is eligible for Evening shows")
```

# Step 5
1. Add an else clause to my if statement and 
2. print "User is not eligible for Evening shows" inside the else body. This will print only when the condition in the if statement is false.

```python
if age >= 21:
    print('User is eligible for Evening shows')
else:
    print("User is not eligible for Evening shows")
```

# Step 6
1. Create a variable named is_member to indicate whether the user is a member and set its value to True.
2. Create another variable named is_weekend to indicate whether the movie show is on a weekend and set its value to False. Do not surround the value with quotes.

```python
is_member = True
is_weekend = False
```

# Step 7
The user qualifies for a membership discount if they are a member.

```python
discount = 0
```

# Step 8
1. Create an if statement to check if is_member is truthy. 
2. Inside the body of the if statement, update the discount value to 3 
3. Print "User qualifies for membership discount" to the terminal.

```python
if is_member:
    discount = 3 
    print("User qualifies for membership discount")
```

# Step 9
1. Handle the case when the user does not qualify for a membership discount.
2. Display the updated value of discount.

```python
if is_member:
    discount = 3
    print('User qualifies for membership discount')
else:
    print("User does not qualify for membership discount")
print("Discount:", discount)
```

# Step 10 
The membership discount should only apply to members if their age is greater than or equal to 21.

```python
if is_member and age >= 21:
    discount = 3
    print('User qualifies for membership discount')
else:
    print('User does not qualify for membership discount')
print('Discount:', discount)
```

# Step 11
Change the value of the is_member variable to False as the user is not a member.

```python
is_member = False
is_weekend = False

discount = 0
if is_member and age >= 21:
    discount = 3
    print('User qualifies for membership discount')
else:
    print('User does not qualify for membership discount')
print('Discount:', discount)
```

The discount value now remains 0, because both condition (membership status and age equal or greater than 21) must be satisfied for the discount to apply.

# Step 12
1. Add extra charges to apply to the movie ticket on weekends.
2. Check if is_weekend is truthy.
3. Update the extra_charges value to 2 
4. Print "Extra charges will be applied" in the terminal.

```python
extra_charges = 0
if is_weekend:
    extra_charges = 2
    print("Extra charges will be applied")
```

# Step 13
1. Add an else clause to my if statement and print "No extra charges will be applied" inside the else body.
2. Display a message that shows "Extra charges:", followed by the updated value of extra_charges and check the output in the terminal.

```python
extra_charges = 0
if is_weekend:
    extra_charges = 2
    print('Extra charges will be applied')
else:
    print("No extra charges will be applied")
print("Extra charges:", extra_charges)
```

# Step 14
Extra charges should also apply if the show is in the evening.

```python
extra_charges = 0
if is_weekend or show_time == "Evening":
    extra_charges = 2
    print('Extra charges will be applied')
else:
    print('No extra charges will be applied')
print('Extra charges:', extra_charges)
```

# Step 15
Check if the user satisfies the conditions to book a movie ticket. Users with age 21 or above can always book tickets without any restrictions.

```python
if age >= 21:
    print("Ticket booking condition satisfied")
else:
    print("Ticket booking failed due to restrictions")
```

# Step 16
1. Users between 18 and 21 can book tickets, but only when the show_time is not Evening.
2. Update the condition of the if age >= 21: line.
3. Do not create new variables.

```python
if age >= 21 or age >= 18 and show_time != "Evening":
    print('Ticket booking condition satisfied')
else:
    print('Ticket booking failed due to restrictions')
```

# Step 17
Users who are at least 18 years old but younger than 21 can book evening shows if they are members.

```python
if age >= 21 or age >= 18 and (show_time != 'Evening' or is_member):
    print('Ticket booking condition satisfied')
else:
    print('Ticket booking failed due to restrictions')
```

Note: 
1. When multiple logical operators are used in an if statement, "and" has a higher precedence than "or". This means conditions joined with "and" are grouped together, as if they were wrapped in parentheses. 
2. That's why I can use parentheses () to change how conditions are grouped. Such as show_time != 'Evening' and is_member above with the "or" operator.

# Step 18
Calculate service charges based on the type of seat the user has selected.

```python
if age >= 21 or age >= 18 and (show_time != 'Evening' or is_member):
    print('Ticket booking condition satisfied')

    service_charges = 0
    if seat_type == "Premium":
        service_charges = 5
else:
    print('Ticket booking failed due to restrictions')
```

# Step 19
Inside the body of the outer if statement:
1. Add an else clause to the nested if seat_type == 'Premium': statement
2. Update the service_charges value to 1 inside the else body.

```python
if age >= 21 or age >= 18 and (show_time != 'Evening' or is_member):
    print('Ticket booking condition satisfied')

    service_charges = 0
    if seat_type == 'Premium':
        service_charges = 5
    else:
        service_charges = 1    
else:
    print('Ticket booking failed due to restrictions')
```

# Step 20
1. Inside the body of the outer if statement, add an elif clause between the if seat_type == 'Premium': and else: lines and check if seat_type is equal to Gold. Inside the body of the elif clause, update the value of service_charges to 3.
2. Display a message that shows "Service charges:" followed by the updated value of service_charges. Then, check the output in the terminal.

```python
if age >= 21 or age >= 18 and (show_time != 'Evening' or is_member):
    print('Ticket booking condition satisfied')

    service_charges = 0
    if seat_type == 'Premium':
        service_charges = 5
    elif seat_type == "Gold":
        service_charges = 3
    else:
        service_charges = 1
    print("Service charges:", service_charges)
else:
    print('Ticket booking failed due to restrictions')
```

# Step 21
Calculate the final price of the movie ticket by adding the extra charges and service charges to the base price, and then subtracting the discount.

```python
if age >= 21 or age >= 18 and (show_time != 'Evening' or is_member):
    print('Ticket booking condition satisfied')

    service_charges = 0
    if seat_type == 'Premium':
        service_charges = 5
    elif seat_type == 'Gold':
        service_charges = 3
    else:
        service_charges = 1
    print('Service charges:', service_charges)

    final_price = base_price + extra_charges + service_charges - discount
    print("Final price of ticket:", final_price)
else:
    print('Ticket booking failed due to restrictions')
```

## Result
The 21-year-old customer is eligible to book a ticket, not a member, and has to pay the final price of 20 dollars. It's from the base price (15 dollars), service charges for the gold seat type (3 dollars), and extra charges for the evening show time (2 dollars), with no membership discount.
