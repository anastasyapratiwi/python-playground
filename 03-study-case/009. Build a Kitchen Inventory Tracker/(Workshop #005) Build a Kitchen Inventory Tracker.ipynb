When a function hits a return statement, it stops executing immediately and passes the returned value back to where it was called.

# Step 9
1. When I do have enough eggs, the if statement above would turn to False and continue running the rest of the code.
2. I need to write code that records the usage of the egg(s) and subtracts automatically from the available eggs.

```python
def use_eggs(available_eggs, eggs_to_use):
    if eggs_to_use > available_eggs:
        print('The kitchen does not have enough eggs.')
        return available_eggs

    print(f'{eggs_to_use} egg(s) used out of {available_eggs} available.')
    return available_eggs - eggs_to_use 
```

# Step 15
Now, my make_fried_egg function checks whether an egg is available and updates its local egg count when one is used. Next, return that count so the caller can use it.

```python
def make_fried_egg(available_eggs):
    has_enough_eggs = available_eggs >= 1

    if has_enough_eggs:
        available_eggs = use_eggs(available_eggs, 1)
        print('Made a fried egg. Yummy!')
    else:
        print('Could not make a fried egg. Not enough eggs!')

    return available_eggs 
```

Note:
1. I just realized by trial and error myself that return has to be in the same indentation as the if-else statement.
2. When the indentation of return is the same as the def function, the terminal will show that it's outside of the function and can't process it.

Note:

2. I have to overwrite the global available_eggs variable with the new number of eggs returned by my make_fried_egg function to update my actual kitchen inventory.
