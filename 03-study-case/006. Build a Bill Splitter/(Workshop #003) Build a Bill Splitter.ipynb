### Subchapter: Build a Bill Splitter (Workshop)

This 8-part workshop is the process of making a tool that calculates how much each person owes after adding meal costs and a tip. A helpful feature after dining with friends, perhaps? Some even use this to go dutch on the first date. I remember they used to have this on LINE and it was a hit!

# Step 1
Keep track of the total amount as costs are added

```python
running_total = 0
```

# Step 2
Create an account for the number of people sharing the bill. Store this value in a variable. It will be used later to calculate the final split.

```python
num_of_friends = 4
```

# Step 3
Add the cost to each course. In this example, I include cents (as in dollars).

```python
appetizers = 37.89
main_courses = 57.34
desserts = 39.39
drinks = 64.21
```

# Step 4
1. Use the += operator once to add appetizers, main_courses, desserts, and drinks to running_total.
2. Use print() to display the string Total bill so far: followed by a space and the value of running_total.

```python
running_total += appetizers + main_courses + desserts + drinks
print("Total bill so far:", running_total)
```

The output has more decimal digits than expected because numbers are stored in binary, and many decimal values cannot be represented exactly in this format, which leads to rounding errors.

# Step 5
Since the service was excellent, the group decides to leave a 25% tip.

```python
tip = running_total * 0.25
print("Tip amount:", tip)
```

# Step 6
Add the tip to the running_total to find the final bill amount

```python
running_total += tip
print("Total with tip:", running_total)
```

# Step 7
Determine how much each person owes by dividing the total bill by the number of friends.

```python
final_bill = running_total / num_of_friends
print("Bill per person:", final_bill)
```

# Step 8
Since money is typically represented with two decimal places, I should round the final result.

```python
each_pays = round(final_bill, 2)
print("Each person pays:", each_pays)
```

The bill splitter workshop is complete!
