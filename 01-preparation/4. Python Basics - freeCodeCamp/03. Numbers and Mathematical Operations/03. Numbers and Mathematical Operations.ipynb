# A. How Do You Work With Integers and Floating Point Numbers? 

## 1. With Integers 

```python
my_int_1 = 56
my_int_2 = -4
print(type(my_int_1))  # <class 'int'>
print(type(my_int_2))  # <class 'int'>
```

### a. Addition operation with integers

```python
my_int_1 = 56
my_int_2 = 12
sum_ints = my_int_1 + my_int_2
print("Integer Addition:", sum_ints)  # Integer Addition: 68
```

### b. Substraction with integers

```python
my_int_1 = 56
my_int_2 = 12

# Substraction
diff_ints = my_int_1 - my_int_2
print("Integer Substraction:", diff_ints)  # Integer Substraction: 44
```

### c. Multiplication operation with integers

```python
my_int_1 = 12
my_int_2 = 4

# Multiplication
product_ints = my_int_1 * my_int_2
print("Integer Multiplication:", product_ints)  # Integer Multiplication: 48
```

### d. Division operation with integers

```python
my_int_1 = 56
my_int_2 = 12

# Division
div_ints = my_int_1 / my_int_2
print("Division:", div_ints)  # Division: 4.666666666666667
```

## 2. With Floats
Floats are positive or negative numbers with decimal points, like 3.14, -0.5, or 0.0.

### Example code

```python
my_float_1 = -12.0
my_float_2 = 4.9

print(type(my_float_1)) # <class 'float'>
print(type(my_float_2)) # <class 'float'>
```

### a. Addition

```python
my_float_1 = 5.4
my_float_2 = 12.0

float_addition = my_float_1 + my_float_2
print('Float Addition: ', float_addition) # Float Addition: 17.4
```

### b. Substraction

```python
my_float_1 = 5.4
my_float_2 = 12.0

float_substraction = my_float_2 - my_float_1
print('Float Substraction: ', float_substraction) # Float Substraction: 6.6
```

### c. Multiplication

```python
my_float_1 = 5.4
my_float_2 = 12.0

float_multiplication = my_float_2 * my_float_1
print('Float Multiplication: ', float_multiplication) # Float Multiplication: 64.80000000000001
```

### d. Division

```python
my_float_1 = 5.4
my_float_2 = 12.0

float_division = my_float_2 / my_float_1
print('Float Division: ', float_division) # Float Division: 2.222222222222222
```

### e. Add an integer and a float (Result: a float)

```python
my_int = 56
my_float = 5.4

sum_int_and_float = my_int + my_float

print(sum_int_and_float) # 61.4
print(type(sum_int_and_float)) # <class 'float'>
```

If I mix integers and floats, even with other basic arithmetic operations like subtraction, multiplication, and division, the result is a float.

## 3. More complex arithmetic calculations

### a. Modulo operator (%)
It returns the remainder when the value on the left is divided by the value on the right

```python
my_int_1 = 56
my_int_2 = 12

my_float_1 = 5.4
my_float_2 = 12.0

mod_ints = my_int_1 % my_int_2
mod_floats = my_float_2 % my_float_1

print('Integer Modulo: ', mod_ints) # Integer Modulo: 8
print('Float Modulo: ', mod_floats) # Float Modulo: 1.1999999999999993
```

### b. Floor division (//)
It divides two numbers and returns the greatest integer less than or equal to the result. 

```python
my_int_1 = 56
my_int_2 = 12

my_float_1 = 5.4
my_float_2 = 12.0

floor_div_ints = my_int_1 // my_int_2
floor_div_floats = my_float_2 // my_float_1

print('Integer Floor Division: ', floor_div_ints) # Integer Floor Division: 4
print('Float Floor Division: ', floor_div_floats) # Float Floor Division: 2.0
```

### c. Exponentiation (**)
It raises a number to the power of another

```python
my_int_1 = 56
my_int_2 = 12

my_float_1 = 5.4
my_float_2 = 12.0

exp_ints = my_int_1 ** my_int_2
exp_floats = my_float_1 ** my_float_2

print('Integer Exponentiation: ', exp_ints) # Integer Exponentiation: 951166013805414055936
print('Float Exponentiation: ', exp_floats) # Float Exponentiation: 614787626.1765089
```

Sometimes, the result of an operation with floats has more decimal digits than expected. For example, the sum 0.1 + 0.2 equals 0.30000000000000004 instead of 0.3.

Numbers are stored in binary format, and some fractions can't be represented exactly in binary. As a result, they are stored as finite approximations, in the same way the fraction 1/3 cannot be represented with a finite number of digits in decimal and is truncated after a certain number of its infinite digits (0.33333...).

This leads to small rounding errors and I'll learn how to round the number.

## 4. Built-in functions

### a. float()
It converts an integer into a float.

```python
my_int_1 = 56
my_float_1 = float(my_int_1)

print(my_float_1) # 56.0
print(type(my_float_1)) # <class 'float'>
```

### b. int()
It converts a float into an integer

```python
my_float = 12.92563
my_int = int(my_float)

print(my_int) # 12
print(type(my_int)) # <class 'int'>
```

### Example for both float() and int()

```python
my_str_int = '45'
my_str_float = '7.8'
converted_int = int(my_str_int)
converted_float = float(my_str_float)

print(converted_int, type(converted_int)) # 45 <class 'int'>
print(converted_float, type(converted_float)) # 7.8 <class 'float'>
```

## 5. Other methods

### a. round()
Rounds a number to the specified number of decimal places. By default this function rounds to the nearest integer, and returns a whole number with no decimal places

```python
my_int_1 = 4.798
my_int_2 = 4.253

rounded_int_1 = round(my_int_1)
rounded_int_2 = round(my_int_2, 1) # For one decimal

print(rounded_int_1) # 5
print(rounded_int_2) # 4.3
```

### b. abs()
Returns the absolute value of a number

```python
num = -15

absolute_value = abs(num)
print(absolute_value) # 15
```

### c. pow()
Raises a number to the power of another or performs modular exponentiation

```python
result_1 = pow(2, 3) # Equivalent to 2 ** 3
print(result_1) # 8

result_2 = pow(2, 3, 5) # Equivalent to (2 ** 3) % 5
print(result_2) # 3
```

# B. How Do Augmented Assignments Work?
Augmented assignment applies an operation to a variable and stores the result back in the same variable, all in one step.

## Basic syntax
variable <operator>= value

## 1. Addition (+=)
Let's compare the codes below.

### a.  Using augmented assignment to add 5

```python
my_var = 10
my_var += 5

print(my_var) # 15
```

### b. Without augmented assignment

```python
my_var = 10
my_var = my_var + 5

print(my_var) # 15
```

Augmented assignment lets me update a variable value without repeating the variable name. In turn, this reduces potential errors that might arise from a typo or something similar.

Another operators also have their own version of augmented assignment and I'll learn it.

## 2. Substraction (-=)
Subtracts the right operand from the left variable and stores the difference in the left variable

```python
count = 14
count -= 3

print(count) # 11
```

## 3. Multiplication (*=)
Multiplies the left variable by the right operand and stores the product back in the left variable

```python
product = 65
product *= 7

print(product) # 455
```

## 4. Division (/=)
Divides the left variable by the right and stores the result back in the left variable

```python
price = 100
price /= 4

print(price) # 25.0
```

## 5. Floor division (//=)
Divides the left variable by the right and stores the result back in the left variable

```python
total_pages = 23
total_pages //= 5

print(total_pages) # 4
```

## 6. Modulo assignment operator (%=)
Computes the remainder of the left variable divided by the right and stores it back in the left variable

```python
bits = 35
bits %= 2

print(bits) # 1
```

## 7. Exponentiation assignment operator (**=)
Raises the left variable to the power of the right and stores the result back in the left variable 

```python
power = 2
power **= 3

print(power) # 8
```

## 8. Augmented assignment operators with strings

### a. Addition assignment + strings

```python
greet = 'Hello'
greet += ' World'

print(greet) # Hello World
```

### b. Multiplication assignment to repeat a string

```python
greet = 'Hello'
greet *= 3

print(greet) # HelloHelloHello
```

### c. Other augmented assignments with strings = TypeError

```python
greet = 'Hello'
greet -= ' World'

print(greet) # TypeError: unsupported operand type(s) for -=: 'str' and 'str'
```

```python
greet = 'Hello'
greet /= ' World'
print(greet) # TypeError: unsupported operand type(s) for /=: 'str' and 'str'
```
