# A. What Are Strings and What Is String Immutability?
A string is a sequence of characters surrounded by either single or double quotation marks

## Example code

```python
my_str_1 = 'Hello'
my_str_2 = "World"
```

## Multi-line string example

```python
my_str_3 = """Multiline
string"""
my_str_4 = '''Another
multiline
string'''
```

## If the strings contains either single or double quotation marks

### a. Use the opposite kind of quotes

```python
msg = "It's a sunny day"
quote = 'She said, "Hello World!"'
```

### b. Use backlash (\\) for the same quotation marks

```python
msg = 'It\'s a sunny day'
quote = "She said, \"Hello!\""
```

## in operator (Check if a string contains one or more characters)
It returns a boolean that specifies whether the character or characters exist in the string or not

```python
my_str = 'Hello world'

print('Hello' in my_str) # True
print('hey' in my_str) # False
print('hi' in my_str) # False
print('e' in my_str) # True
print('f' in my_str) # False
```

## Indexing

### len(): Get the length of a string

```python
my_str = 'Hello world'
print(len(my_str)) # 11
```

### []: Access a character by its index
The index is zero-based, meaning that the index of the first character of a string is 0, the index of the second character is 1, and so on. 

```python
my_str = "Hello world"
print(my_str[0]) # H
print(my_str[6]) # w
```

### Negative indexing (from the last character)

```python
my_str = 'Hello world'
print(my_str[-1]) # d
print(my_str[-2]) # l
```

## Values in Python
1. In Python, values can be mutable or immutable. A mutable value can be changed after it is created, while an immutable value cannot.
2. I can point a variable at a new value (reassignment), but I can't change an immutable value itself by adding, removing, or replacing any of its elements.
3. Strings are immutable in Python
4. Integers, floats, and booleans are also immutable.

### Assign a different string to a variable (immutable)

```python
greeting = 'hi'
greeting = 'hello'
print(greeting) # hello
```

### Don't direct modification of a string

```python
greeting = 'hi'
greeting[0] = 'H' # TypeError: 'str' object does not support item assignment
```

# B. What Are String Concatenation and String Interpolation?

## 1. String Concatenation (+)
I can combine multiple strings together with the plus (+) operator.

```python
my_str_1 = 'Hello'
my_str_2 = "World"

str_plus_str = my_str_1 + ' ' + my_str_2
print(str_plus_str) # Hello World
```

## 2. Repeating Strings (*)
I can also repeat a string by multiplying it with an integer using the * operator. The string is repeated the specified number of times.

```python
sound = 'ha'
repeated_sound = sound * 3
print(repeated_sound) # hahaha
```

## 3. Concatenating Strings with Numbers

### a. Directly (TypeError)
Concatenation only works with strings. If I try to concatenate a string with a number, I'll get a TypeError

```python
name = 'John Doe'
age = 26
name_and_age = name + age
print(name_and_age) # TypeError: can only concatenate str (not "int") to str
```

This happens because Python does not automatically convert other data types like integers into strings when I concatenate them.

### b. str(): convert the value into a string
Python requires all elements to be strings before it can concatenate them. To fix that, I can convert the number into a string with the built-in str() function, which returns the string representation of the given value without modifying the original value


```python
name = 'John Doe'
age = 26

name_and_age = name + str(age)
print(name_and_age) # John Doe26
```

### c. Concatenation and assignment in one step (+=)

```python
name = 'John Doe'
age = 26

name_and_age = name # Start with the name
name_and_age += str(age) # Append the age as string

print(name_and_age) # John Doe26
```

## 4. String Interpolation (f-strings)
Python has a category of string called f-strings (short for formatted string literals). The function is to insert variables and expressions into a string like: f'{}'. Don't convert non-string types with the str() function. These variables are converted under the hood into a string during the interpolation process.

```python
name = 'John Doe'
age = 26
name_and_age = f'My name is {name} and I am {age} years old'
print(name_and_age) # My name is John Doe and I am 26 years old
```

```python
num1 = 5
num2 = 10
print(f'The sum of {num1} and {num2} is {num1 + num2}') # The sum of 5 and 10 is 15
```

# C. What Is String Slicing and How Does It Work?
I previously learned about indexing []

```python
my_str = "Hello world"

print(my_str[0]) # H
print(my_str[6]) # w
print(my_str[-1]) # d
```

## String Slicing
1. Lets me extract a portion of a string or work with only a specific part of it.
2. Basic syntax: string[start:stop]

## 1. Extract characters (separate with a colon) 
If I want to extract characters from a certain index to another, I just separate the start and stop indices with a colon

```python
my_str = 'Hello world'
print(my_str[1:4]) # ell
```

Note: the stop index is non-inclusive, so [1:4] just extracted the characters from index 1, and up to, but not including, the character at index 4.

## 2. Omit to default [0]
I can also omit the start and stop indices, and Python will default to 0 or the end of the string, respectively.

### a. Omit the start index

```python
my_str = 'Hello world'
print(my_str[:7]) # Hello w
```

This extracts everything from index 0 up to (but not including), the character at index 7.

### b. Omit the stop index

```python
my_str = 'Hello world'
print(my_str[8:])
```

This extracts everything from the character at index 8 until the end of the string.

#### Note: Slicing doesn't modify the value of the string

```python
my_str = 'Hello world'
print(my_str[8:]) # rld
print(my_str) # Hello world
```

### c. Omit the start and stop indices
To extract the whole string

```python
my_str = 'Hello world'
print(my_str[:]) # Hello world
```

## 3. Optional: step parameter
1. To specify the increment between each index in the slice.
2. Basic syntax: string[start:stop:step]

### a. Example
The slicing starts at index 0, stops before 11, and extracts every second character.

```python
my_str = 'Hello world'
print(my_str[0:11:2]) # Hlowrd
```

### b. Reverse a string
Set the step to -1, omit both start and stop indices

```python
my_str = 'Hello world'
print(my_str[::-1]) # dlrow olleH
```

# D. What Are Some Common String Methods? 
A method is a function that I call on a value. To call a string method, I write the string or its variable name followed by a dot and the method call.

## 1. .upper(): Converts all characters to uppercase

```python
my_str = 'hello world'
uppercase_my_str = my_str.upper()
print(uppercase_my_str) # HELLO WORLD
```

## 2. .lower(): Converts all characters to lowercase

```python
my_str = 'Hello World'
lowercase_my_str = my_str.lower()
print(lowercase_my_str) # hello world
```

## 3. .strip()
1. Returns a new string with the specified leading and trailing characters removed. 
2. If no argument is passed it removes leading and trailing whitespace.

```python
my_str = '  hello world  '
trimmed_my_str = my_str.strip()
print(trimmed_my_str) # "hello world"
```

## 4. .replace(old, new)
Returns a new string with all occurrences of old replaced by new.

```python
my_str = 'hello world'
replaced_my_str = my_str.replace('hello', 'hi')
print(replaced_my_str) # hi world
```

## 5. .split(separator)
1. Splits a string on a specified separator into a list of strings. 
2. A list groups values between square brackets. If no separator is specified, split() splits on whitespace.

```python
my_str = 'hello world'
split_words = my_str.split()
print(split_words) # ['hello', 'world']
```

## 6. .join()
Joins the strings in a collection into a single string with a separator.

```python
my_list = ['hello', 'world']
joined_my_str = ' '.join(my_list)
print(joined_my_str) # hello world
```

## 7. .startswith(prefix)
Returns a boolean indicating if a string starts with the specified prefix.

```python
my_str = 'hello world'
starts_with_hello = my_str.startswith('hello')
print(starts_with_hello) # True
```

## 8. .endswith(suffix)
Returns a boolean indicating if a string ends with the specified suffix.

```python
my_str = 'hello world'
ends_with_world = my_str.endswith('world')
print(ends_with_world) # True
```

## 9. .find(substring)
Returns the index of the first occurrence of substring, or -1 if it doesn't find one.

```python
my_str = 'hello world'
world_index = my_str.find('world')
print(world_index) # 6
```

## 10. .count(substring)
Returns the number of times a substring appears in a string.

```python
my_str = 'hello world'
o_count = my_str.count('o')
print(o_count) # 2
```

## 11. .capitalize()
Returns a new string with the first character capitalized and the other characters lowercased.

```python
my_str = 'hello world'
capitalized_my_str = my_str.capitalize()
print(capitalized_my_str) # Hello world
```

## 12. .isupper()
Returns True if all letters in the string are uppercase and False if not.

```python
my_str = 'hello world'
is_all_upper = my_str.isupper()
print(is_all_upper) # False
```

## 13. .islower()
Returns True if all letters in the string are lowercase and False if not.

```python
my_str = 'hello world'
is_all_lower = my_str.islower()
print(is_all_lower) # True
```

## 14. .title()
Returns a new string with the first letter of each word capitalized.

```python
my_str = 'hello world'
title_case_my_str = my_str.title() 
print(title_case_my_str) # Hello World
```
