# A. How Do Functions Work in Python?
Aside from print(), there are another built-in functions. They are reusable pieces of code that run when I call them.

## 1. input()
It lets me prompt the user for input. I have to type the output I want, so the process time might differ from time to time. 

```python
name = input('What is your name?') # User types "Kolade" and presses Enter
print('Hello', name) # Output: Hello Kolade
```

## 2. int()
I've learned this before, even completed a workshop where I implemented this function! It converts a number, a boolean, or a string representing an integer into the real integer.

```python
print(int(3.14)) # 3
print(int('42')) # 42
print(int(True)) # 1
print(int(False)) # 0
```

## 3. def
To write my own custom function. The structure is the def keyword, followed by the name I want to give my function, a pair of parentheses, and a colon. Then on a new line, I write the code my function should run. The code the function runs is also referred to as the function's body.

### a. Function named "hello"

```python
def hello():
  print('Hello World')

hello() # Hello World
```

Python relies on indentation to determine which groups of statements belong together. These groups of statements are called code blocks.

### b. Function (sum of two numbers)

```python
def calculate_sum(a, b):
  print(a + b)
```

The calculate_sum function has a and b in its parentheses, separated by a comma. Those are called parameters. I think of parameters as placeholder variables that act as "slots" for the values I pass into functions when I call them.

## 4. Arguments
To use the parameters, I have to pass in "arguments". Arguments are the values I pass to a function when I call it.

### a. Call the function 
Call the calculate_sum function to sum together the numbers 3 and 1 

```python
calculate_sum(3, 1) # 4
```

### b. Without the correct number (TypeError)

```python
calculate_sum()
```

## 5. return
Functions also use a special return keyword to exit the function and return a value. If I don't explicitly use return, Python will return None by default.

### a. Without return (None)

```python
def calculate_sum(a, b):
  print(a + b)

my_sum = calculate_sum(3, 1) # 4
print(my_sum) # None
```

1. The calculate_sum function prints the sum of a and b, but it doesn't return anything explicitly. So when I assign its result to my_sum, the value is actually None. 
2. None is a special value that represents the absence of a value. It is the only value of the NoneType data type. None is immutable and falsy, which means it cannot be changed and evaluates to False in a boolean context.

### b. With return

```python
def calculate_sum(a, b):
  return a + b

my_sum = calculate_sum(3, 1)
print(my_sum) # 4
```

The calculate_sum function returns the sum of a and b, which gets stored in my_sum.

# B. What Is Scope in Python and How Does It Work?
1. Scope determines where I can use a variable in my code.
2. For now, my focus is on local and global scope.
3. A variable created outside a function has global scope. I can use it both inside and outside functions.
4. A variable created inside a function has local scope. I can only use it inside that function. Function parameters are local variables too.

## Example 1 (local and global scope)

```python
tax_rate = 0.1

def calculate_tax(price):
  tax = price * tax_rate
  return tax

print(calculate_tax(50)) # 5.0
print(tax_rate) # 0.1
print(tax) # NameError: name 'tax' is not defined
```

1. The tax_rate variable is global because it is created outside the function. The calculate_tax function can read it, and the second print() call can read it too.
2. The price parameter and the tax variable are local to calculate_tax. They are available while that function runs, but not outside it. The last print() call raises a NameError because tax is not defined in the global scope.

## Example 2

```python
discount = 0.2

def apply_discount(price):
  amount = price * discount
  return price - amount
```

The variable that has global scope in the code is discount.

## Example 3

```python
def greet():
  message = 'Hello!'
  print(message)

greet()
print(message)
```

The last line of the code raises NameError because message is defined in the local scope. It's not within the def function and the last print function can't read it.
