# A. How Do You Declare Variables and What Are Naming Conventions to Name Variables?
To create a variable, I write its name on the left, followed by the assignment (=) operator and the value I want to store on the right. 

```python
# Example to create name and age variables:
name = 'John Doe'
age = 25
```

The variable name holds the value 'John Doe'. This value is a string, which is a series of characters used to represent text. Strings are written with single or double quotes, for example 'Hello' or "Hello"

## Important rules of naming variables in Python
1. Variable names can only start with a letter or an underscore (_), not a number.
2. Variable names can only contain alphanumeric characters (a-z, A-Z, 0-9) and underscores (_).
3. Variable names are case-sensitive: age, Age, and AGE are all considered unique.
4. Variable names cannot be one of Python's reserved keywords such as if, class, or def.

## What happens if I break any of these rules:
My Python program would crash and produce a syntax error as the output

```python
5variable_name = 5
```

## Some common naming conventions for variables in Python

1. Variable names should be in lowercase, separated by an underscore (snake case)

```python
my_variable_name = 'freeCodeCamp'
```

2. Use descriptive names for variables

Example: if I want to save a user's age as a variable, user_age is better than age or an abbreviation like ua

```python
user_age = 30
```

3. Avoid using single-letter variable names

Single-letter names are common in Python, but they should be avoided because they communicate no purpose or meaning

```python
x = 56 # What do you mean by x?
```

4. The pound symbol (#) and the text that follows it form a comment

Comments let me add notes and explanations to my code. The Python programming language ignores everything after the # symbol on that line

```python
# This is a single-line comment
```

5. Multi-line comments can be created by using consecutive single-line comments

```python
# This is a
# multi-line
# comment
```

Tips: don't use comments to explain what the variable names mean. I can use comments to explain my code, leave reminders for myself, or clarify why a line exists. Comments are especially helpful when I'm learning or working in teams.

# B. How Does the Print Function Work?
The print() function sends text and other values to the terminal. 

## A common first program display

```python
print('Hello world!') # Hello world!
```

Note: Strings as a sequence of characters surrounded by either single (') or double (") quotation marks.

## Show multiple values or arguments
The string 'Hello world!' is an argument passed to the print function. I can also use the print function to show multiple values, or arguments, at once by separating them with commas.

```python
print('My favorite colors are', 'blue', 'green', 'red')
# Output: My favorite colors are blue green red
```

Python automatically adds a space between each item when I separate them with commas. This is helpful when I want to print several pieces of information together.

# C. What Are Common Data Types in Python?
A data type describes the kind of value a variable holds.  Python determines the type from the value assigned to the variable.

## Example

```python
name = 'John Doe' # Python knows this is a string
age = 25 # Python knows this is an integer
```

## Assigned a variable to a different type

```python
age = 25
age = 'Twenty-five'
```

After the second assignment, age contains a string instead of an integer.

## 4 basic data types I need to know

### 1. Integer (whole number without decimals)

```python
my_integer_var = 10
print('Integer:', my_integer_var) # Integer: 10
```

### 2. Float (number with a decimal point)

```python
my_float_var = 4.50
print('Float:', my_float_var) # Float 4.5
```

### 3. String (characters enclosed in single or double quotation marks)

```python
my_string_var = 'hello'
print('String:', my_string_var) # String: hello
```

### 4. Boolean (True or False)

```python
my_boolean_var = True
print('Boolean:', my_boolean_var) # Boolean: True
```

# D. How Do the type() and isinstance() Functions Work?
As I build out my programs, I need to learn how to view the type of a variable.

## 1. type() function
To see what type developer is, I can use the type() function

```python
developer = 'Devin'
print(type(developer)) # <class 'str'> means String
```

The output of <class 'str'> means that developer is a string type.

### Complete the data types

```python
my_integer_var = 10
print(type(my_integer_var)) # <class 'int'>

my_float_var = 4.50
print(type(my_float_var)) # <class 'float'>

my_string_var = 'hello'
print(type(my_string_var)) # <class 'str'>

my_boolean_var = True
print(type(my_boolean_var)) # <class 'bool'>
```

## 2. isinstance() function
Usage: when I need to verify that a particular variable is a specific type before performing operations on it. 

### a. Error when trying to perform mathematical operations

```python
account_balance = '12'
account_balance / 2
```

### b. See if the data type is an integer
Since the output shows errors, I try to make sure if the data I input is an integer

```python
account_balance = '12'
isinstance(account_balance, int) # False
```

The isinstance() function lets me check if a variable matches a specific data type. It takes a value and the type I want to check it against, then returns a boolean. In this case, since account_balance is a string, it will return False.

### c. Check for multiple types at once

```python
account_balance = 12
isinstance(account_balance, (int, float)) # True
```

The account_balance is an integer so isinstance() returns True. If account_balance were 12.0, isinstance() would still return True because I'm checking for integers or floats.
