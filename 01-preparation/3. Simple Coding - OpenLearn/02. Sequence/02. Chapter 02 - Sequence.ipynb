### Activity 2
Put double-quotes around the variable name in the second line. Run the program. Try to explain what happened before reading further.

```python
result = 3 + 7
print("The sum of 3 and 7 is:"), "result"
```

#### Explanation:  "result", "7", result and 7 are different things.

2. result: variable
3. 7: number

```python
# The correct code:
result = 3 + 7
print("The sum of 3 and 7 is:"), result
```
