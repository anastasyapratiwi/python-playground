### Activity 7
Calculate and print how many items were ordered, i.e. 5 in the example above. First change the algorithm, then translate it to code. Hint: you need to count the items in the list, their prices are irrelevant.

```python
# - a. Using for loop -
# Let the items be a list of the prices of the food and drinks ordered.
items = [4.35, 2, 19.95, 22.70, 5]
# Set the counter to zero
total_items = 0
# Input the logic with for loop
for item in items:
  total_items +=1 # Adds 1 to the counter
# Print the result
print(f"Total items ordered: {total_items}")
```

```python
# - b. Using len() -
# Let the items be a list of the prices of the food and drinks ordered.
items = [4.35, 2, 19.95, 22.70, 5]
# Calculate the total of items were ordered
total_items = len(items)
# Print the result
print(f"Total items ordered: {total_items}")
```
