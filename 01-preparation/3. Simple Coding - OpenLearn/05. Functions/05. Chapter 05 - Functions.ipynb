### The sum function
Calculating the sum of a list of numbers. In this case, the sum function takes a list of numbers as input and returns a single number as output.

```python
items = [4.35, 2, 19.95, 22.70, 5]
expenses = sum(items)
print("Food and drinks:"), expenses
```

### The length function
Counting the number of items in a list, which returns the length of the list.

```python
# Change the code above to show the number of ordered items, using the len function.
items = [4.35, 2, 19.95, 22.70, 5]
expenses = len(items)
print("Food and drinks:"), expenses
```

## 5.1 The input function
input() takes a string that it prints on the screen (the user’s prompt), and returns a string with what the user typed on the keyboard until they pressed the RETURN or ENTER key. Input always returns a string, even if the user typed in a number.

```python
# Run the code below and type in your name, followed by RETURN or ENTER.
name = input("Tasya")
print("Nice to meet you,"), name
```

```python
# Run the code below and type in your birth year.
year = input(2001)
print("Your age is"), 2026 - year
```

The computer reports a type error, because the code is mixing two different types of data. It is trying to subtract the string returned by input.

## 5.2 A conversion function
The solution: int() that takes a string -> integer (number). If the string represents a decimal number, int will give an error.


### Example

```python
answer = input("In what year were you born?")
year = int(2001)
print("Your age is"), 2026 - year
```

### Activity 8
Change the restaurant program (repeated below) so that it asks the user for the number of people.

```python
expenses = 54
answer = input("How many people?")
people = int(answer)
if people >= 15:
  percentage = 0.15
elif people > 6:
  percentage = 0.10
else:
  percentage = 0
tip = expenses * percentage
bill = expenses + tip
print("Total bill:"), bill
```
